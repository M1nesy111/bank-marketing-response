# 01 认识数据

这是我们的第一个笔记本。今天只做三件事：

1. 把数据读进来
2. 看看有多少行、多少列
3. 看看买的人占多少

每运行完一格，就看一下下面的结果。

```python
import pandas as pd

# 读数据。注意这份文件用分号分隔，不是逗号，所以要写 sep=";"
df = pd.read_csv("../data/bank-additional-full.csv", sep=";")

df.head()   # 看看前 5 行长什么样
```

## 1. 有多少行、多少列

每一行是一个客户，每一列是一个特征。

```python
print("行数（客户数）：", df.shape[0])
print("列数（特征数）：", df.shape[1])
print("\n所有列名：")
print(df.columns.tolist())
```

## 2. 买的人占多少

`y` 这一列是答案：yes = 买了，no = 没买。

```python
# 数一数 yes 和 no 各有多少个
print(df["y"].value_counts())

# 算比例（normalize=True 表示算百分比）
print("\n占比：")
print(df["y"].value_counts(normalize=True).round(3))
```
